```python
def greet(name):
    return "Hello, " + name

result = greet("Harika")
print(result)
```

```python
students = ["Harika", "Ravi", "Sneha", "Arjun"]
marks = [85, 90, 78, 95]

print("Students:", students)
print("Marks:", marks)
```

```python
def calculate_mean(values):
    return sum(values) / len(values)

mean = calculate_mean(marks)
print("Mean marks:", mean)
```

```python
def calculate_maximum(values):
    return max(values)

maximum = calculate_maximum(marks)
print("Maximum marks:", maximum)
```

```python
def calculate_minimum(values):
    return min(values)

minimum = calculate_minimum(marks)
print("Minimum marks:", minimum)
```

```python
def calculate_percentage(obtained, total):
    if total <= 0:
        return None
    return (obtained / total) * 100

percentage = calculate_percentage(425, 500)
print("Percentage:", percentage, "%")
```

```python
def calculate_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 80:
        return "B"
    elif mark >= 70:
        return "C"
    elif mark >= 60:
        return "D"
    else:
        return "F"

for mark in marks:
    print(mark, "->", calculate_grade(mark))
```

```python
print("Student Marks Analysis")
print("----------------------")
print("Marks:", marks)
print("Mean:", calculate_mean(marks))
print("Maximum:", calculate_maximum(marks))
print("Minimum:", calculate_minimum(marks))
print("Percentage:", calculate_percentage(425, 500), "%")
print("Grade for 85 marks:", calculate_grade(85))
```
